## Abstract

The administration of capital punishment in the United States has long been subject to
scrutiny for racial, geographic, and demographic disparities. This project investigates
what factors are associated with death row outcomes, including execution, resentencing,
exoneration, and commutation, among individuals sentenced to death between 1976
and 2019, using data compiled by The Intercept.

We analyzed demographic variables (race, gender, age at sentencing), geographic variables
(state), and temporal variables (sentencing year) using exploratory data analysis,
correlation analysis, simple and multiple OLS regression, an independent samples t-test,
a log transformation, and a chi-square test of independence.

Our findings show that sentencing year is the strongest predictor of execution outcome,
with cohort execution rates declining significantly over time. State execution rate was the strongest geographic predictor, reflecting
large jurisdictional differences in how the death penalty is applied. Race and gender
showed statistically significant but modest associations. White defendants had higher
execution rates relative to other groups, male defendants were executed at nearly twice
the rate of female defendants, and executed individuals tended to be sentenced at slightly
younger ages than those not executed.

These results are consistent with prior research documenting systemic disparities in
capital punishment and highlight that where and when a person is sentenced matters as
much as individual demographic characteristics. Limitations include substantial missing
data for age at sentencing, the absence of appeal-level and victim
characteristic data, and unresolved cases for recent sentencing cohorts whose outcomes have not yet been recorded.

## Research Question

What demographic, legal, and geographic factors are associated with death row outcomes (execution, commutation, exoneration, or death from other causes) among individuals sentenced to death in the United States since 1976?

## Background and Prior Work

Our interest in research is on what factors are associated with death row outcomes among individuals sentenced to death in the United States since 1976, including demographics, legal factors, and geographic distribution. The death penalty in the United States remains one of the most researched topics of the criminal justice system, with outcomes on death row shaped by more than just legal facts. Since its reestablishment in 1976, thousands of individuals have been sentenced to death, executed, exonerated, had sentences commuted, or died from other causes while on death row<sup><a name="ref3"></a>3</sup>. National data from *Death Row U.S.A.* indicate that death row populations are disproportionately male (98.9 %) and primarily White or Black, with significant variation across states, suggesting the importance of geographic context in analyzing death row outcomes<sup><a name="ref3"></a>3</sup>.

Research on specific jurisdictions has studied how demographic factors, particularly race and ethnicity, influence outcomes after a death sentence. Petrie and Coverdill researched Texas death row cases from 1974 to 2009 and found that the race of both offenders and victims affected the likelihood of execution, while having less impact on sentence relief such as commutation or overturning<sup><a name="ref4"></a>4</sup>. These findings suggest that while executions may reflect demographic disparities, other outcomes are shaped by different factors, including appeals processes and legal context.

The results in studies of high-activity death penalty counties show how race interacts with legal processes. Scott Phillips’ analysis of Harris County, Texas, shows that both defendant and victim race influenced whether a defendant was charged with capital murder and ultimately given the death penalty<sup><a name="ref4"></a>4</sup>. Combined with national data, these studies indicate that death row outcomes arise from the intersection of individual characteristics, legal decision-making, and geographic variation. Understanding these factors is important when addressing questions about individual justice and disparities in the administration of capital punishment.

### Data Overview — The Condemned: Death Sentences in the Modern Era (1976–2019)

- Dataset Name: The Condemned

- Number of observations: 7,335

This dataset, compiled by *The Intercept*, includes all known individuals sentenced to death in active U.S. death penalty jurisdictions from July 2, 1976 (the date of *Gregg v. Georgia*, which reinstated the modern era of capital punishment) through November 30, 2019. Each row represents one individual sentenced to death during the modern death penalty era.

As of 2019:
- 1,448 individuals were executed  
- 2,752 remain on death row  
- 3,135 were removed from death row for reasons other than execution (such as commutations, exonerations, resentencing, death in custody, or release)

This dataset allows us to examine how demographic, geographic, and temporal factors are associated with different death row outcomes across jurisdictions in the modern U.S. death penalty system.

### Variables most relevant to this project

Outcome variables
- executed: Indicates whether the individual was executed.
- volunteer_execution: Indicates whether the individual waived appeals and voluntarily accepted execution.
- commuted: Indicates whether the death sentence was commuted through executive action.
- exonerated: Indicates whether the individual was legally cleared of the charges.
- resentenced: Indicates whether the individual received a non-death sentence after review.
- released: Indicates whether the individual was released from prison.
- died / suicide: Indicates whether the individual died (including suicide) while on death row.
- currently_on_death_row: Indicates whether the individual remained on death row at the time of the dataset’s most recent update.

Demographic variables
- race: Recorded race of the individual, used to examine potential racial disparities in death row outcomes.
- gender: Recorded gender of the individual.
- DOB (date of birth): Used to calculate age at the time of sentencing.

Geographic variables
- state: The U.S. state where the individual was sentenced to death, allowing comparison across jurisdictions.
- county: The county of sentencing, reflecting the local legal and judicial context.

Additional variables
- sentencing_year: The year the death sentence was issued, allowing analysis of trends in capital punishment practices over time.
- current_sentence: Describes the individual’s current legal status after removal from death row or continued incarceration. Possible values include DR (Death Row), LWOP (Life Without Parole), LPP (Life With Possibility of Parole), Y (Term of Years), M (Term of Months), and TIME (Time Served).

### Dataset limitations relevant to this project

1. The dataset only includes individuals who received a death sentence. It does not include people charged with capital crimes who did not receive the death penalty, so disparities in who receives a death sentence cannot be analyzed.

2. The data were compiled from corrections departments, public defenders, advocacy groups, and independent trackers. Reporting standards may vary across states and may introduce inconsistencies.

3. Individuals sentenced earlier have had more time for appeals, commutations, or execution, so recent sentences may appear more often as “currently on death row.”

4. The dataset does not include detailed information about appeals, crime type, victim characteristics, or offense severity, which are important legal factors that may influence sentencing outcomes.

5. The dataset also lacks specific dates for executions, commutations, or exonerations, making it difficult to measure time spent on death row or analyze how quickly different outcomes occur.

#### Distribution of Death Row Outcomes

#### Quantitative Analysis: Correlations Between Key Variables

To explore the relationships between demographic and temporal variables and death row outcomes, we compute point-biserial correlations between each predictor and the binary execution outcome. This provides a quantitative summary of which variables are most strongly associated with whether an individual was ultimately executed.

````python
df['race_clean'] = df['race'].str.strip().str.title().replace(
    {'Asian': 'Other', 'Native American': 'Other', 'Other Or Unknown': 'Other'}
)
race_dummies = pd.get_dummies(df['race_clean'], prefix='race')
df_encoded = pd.concat([df, race_dummies], axis=1)


if df_encoded['executed'].dtype == 'bool':
    df_encoded['executed_num'] = df_encoded['executed'].astype(int)


cols_to_drop = ['state_exec_rate_x', 'state_exec_rate_y', 'state_exec_rate']
df_encoded = df_encoded.drop(columns=[col for col in cols_to_drop if col in df_encoded.columns])


state_exec_rate = df_encoded.groupby('state')['executed_num'].mean().reset_index()
state_exec_rate.columns = ['state', 'state_exec_rate']
df_encoded = df_encoded.merge(state_exec_rate, on='state', how='left')


corr_vars = ['sentencing_year', 'gender_male', 'state_exec_rate',
             'race_Black', 'race_Hispanic Or Latino', 'race_White']

corr_results = {}
for var in corr_vars:
    valid = df_encoded[['executed_num', var]].dropna()
    r, p = stats.pointbiserialr(valid['executed_num'], valid[var])
    corr_results[var] = {'r': round(r, 3), 'p': round(p, 4)}

corr_df = pd.DataFrame(corr_results).T.sort_values('r', ascending=False)
corr_df.index = ['Sentencing Year', 'Gender (Male)', 'State Execution Rate',
                 'Race: Black', 'Race: Hispanic/Latino', 'Race: White']

fig, ax = plt.subplots(figsize=(9, 4))
colors = ['crimson' if r > 0 else 'steelblue' for r in corr_df['r']]
bars = ax.barh(corr_df.index, corr_df['r'], color=colors, alpha=0.8)
ax.axvline(0, color='black', linewidth=0.8)
ax.set_xlabel('Point-Biserial Correlation (r) with Execution Outcome')
ax.set_title('Correlation of Predictor Variables with Execution Outcome')

for bar, (_, row) in zip(bars, corr_df.iterrows()):
    sig = '***' if row['p'] < 0.001 else ('**' if row['p'] < 0.01 else ('*' if row['p'] < 0.05 else ''))
    ax.text(bar.get_width() + 0.003, bar.get_y() + bar.get_height()/2,
            f"r={row['r']:.3f} {sig}", va='center', fontsize=9)

ax.text(0.01, -0.12, '* p<.05  ** p<.01  *** p<.001', transform=ax.transAxes, fontsize=6, color='gray')
plt.tight_layout()
plt.show()

print('\nCorrelation Summary Table:')
print(corr_df.to_string())
````

The correlation analysis shows that sentencing year has the strongest association with execution outcome (r = 0.482, p < .001). This positive correlation reflects the dataset's snapshot nature: individuals sentenced in earlier years (smaller year values) have had more time for their cases to resolve, so the executed population skews toward older cohorts. In other words, a higher sentencing year is associated with lower odds of having been executed yet, not higher. The positive r is a feature of how time affects case resolution, not evidence that more recent sentences lead to more executions.

Gender (Male) shows a small but statistically significant positive correlation with execution (r = 0.074, p < .001), suggesting males are slightly more likely to have been executed. State execution rate also shows a small positive correlation (r = 0.031, p < .01), indicating that individuals sentenced in higher-executing states are modestly more likely to appear as executed in the dataset.

All three racial categories show negative correlations with execution. Race: White has the largest negative correlation (r = −0.189, p < .001), followed by Race: Hispanic/Latino (r = −0.056, p < .001). Race: Black shows a correlation close to zero and is not statistically significant (r = −0.010, p = .37). Overall, these results suggest that demographic variables have relatively small associations with execution outcomes, and the strongest relationship observed in this exploratory analysis is related to sentencing year.

#### Logistic Regression: Predicting Execution Outcome

To quantify how multiple demographic and geographic variables jointly relate to execution outcome, we fit a logistic regression model. This allows us to examine the contribution of each variable while controlling for the others, providing a more interpretable picture of the factors associated with execution.

````python
# Build model dataset
model_vars = ['sentencing_year', 'gender_male', 'state_exec_rate',
              'race_Black', 'race_Hispanic Or Latino', 'race_White']

model_df = df_encoded[model_vars + ['executed_num']].dropna()

X = model_df[model_vars]
y = model_df['executed_num']

# Standardize for interpretable coefficients
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Fit logistic regression
lr = LogisticRegression(max_iter=1000, random_state=42)
lr.fit(X_scaled, y)

# Coefficient plot
coef_labels = ['Sentencing Year', 'Gender (Male)', 'State Exec. Rate',
               'Race: Black', 'Race: Hispanic/Latino', 'Race: White']
coefs = lr.coef_[0]

# Sort by magnitude
sorted_idx = np.argsort(coefs)
coefs_sorted = coefs[sorted_idx]
labels_sorted = [coef_labels[i] for i in sorted_idx]
colors = ['crimson' if c > 0 else 'steelblue' for c in coefs_sorted]

fig, ax = plt.subplots(figsize=(9, 4))
ax.barh(labels_sorted, coefs_sorted, color=colors, alpha=0.8)
ax.axvline(0, color='black', linewidth=0.8)
ax.set_xlabel('Standardized Logistic Regression Coefficient\n(positive = more likely to be executed)')
ax.set_title('Logistic Regression: Variables Associated with Execution Outcome')

# Accuracy
acc = lr.score(X_scaled, y)
ax.text(0.98, 0.02, f'Model Accuracy: {acc:.2%}', transform=ax.transAxes,
        ha='right', fontsize=9, color='gray')

plt.tight_layout()
plt.show()

print(classification_report(y, lr.predict(X_scaled)))
````

The logistic regression model (accuracy: 82.76%) confirms that state execution rate is the strongest predictor (coefficient ≈ 1.20), indicating that geographic differences across states are strongly associated with execution outcomes. Sentencing year has the largest negative coefficient (≈ −0.60), suggesting that individuals sentenced in more recent years are less likely to have been executed within the timeframe captured by the dataset.

Among demographic variables, Race: White has the second largest positive coefficient (≈ 0.35), followed by Race: Black and Gender (Male) (≈ 0.12 each), while Race: Hispanic/Latino is close to zero. The model has difficulty predicting executions because the dataset is imbalanced, with 1,446 executed cases compared to 5,876 non-executed cases. As a result, the model shows lower precision (0.60) and recall (0.38) for the executed class. These results should be interpreted as exploratory associations rather than predictive conclusions.

#### Simple OLS Regression — Sentencing Year vs. Cohort Execution Rate

````python
df['executed_num'] = df['executed'].astype(int)

trend = df.groupby('sentencing_year').agg(
    sentences=('executed_num', 'count'),
    executions=('executed_num', 'sum')
).reset_index()
trend['exec_rate'] = trend['executions'] / trend['sentences']

trend_model = trend[trend['sentencing_year'] <= 2010].copy()
trend_post  = trend[trend['sentencing_year'] > 2010].copy()

m = smf.ols('exec_rate ~ sentencing_year', data=trend_model).fit()
print(m.summary())
````

````python
slope     = m.params['sentencing_year']
intercept = m.params['Intercept']
r2        = m.rsquared
p_val     = m.pvalues['sentencing_year']
ci        = m.conf_int().loc['sentencing_year']
n         = len(trend_model)

x_fit  = np.linspace(1976, 2010, 200)
y_fit  = intercept + slope * x_fit
x_mean = trend_model['sentencing_year'].mean()
ss_x   = ((trend_model['sentencing_year'] - x_mean)**2).sum()
se     = m.bse['sentencing_year']
t_crit = stats.t.ppf(0.975, df=n-2)
se_fit = se * np.sqrt(1/n + (x_fit - x_mean)**2 / ss_x)
y_lo   = y_fit - t_crit * se_fit
y_hi   = y_fit + t_crit * se_fit

fig, ax = plt.subplots(figsize=(10, 5))

ax.scatter(trend_model['sentencing_year'], trend_model['exec_rate'],
           color='steelblue', s=trend_model['sentences']/4,
           alpha=0.7, zorder=3, label='Cohort execution rate (1976–2010)')
ax.scatter(trend_post['sentencing_year'], trend_post['exec_rate'],
           color='lightgray', s=trend_post['sentences']/4,
           alpha=0.7, zorder=3, label='Excluded — cases not yet resolved (post-2010)')
ax.plot(x_fit, y_fit, color='red', linewidth=2, label='OLS regression line')
ax.fill_between(x_fit, y_lo, y_hi, color='red', alpha=0.12, label='95% confidence interval')

stats_text = (f'β = {slope:.4f}  (p < 0.001)\n'
              f'R² = {r2:.3f}\n'
              f'95% CI [{ci[0]:.4f}, {ci[1]:.4f}]')
ax.text(0.98, 0.97, stats_text, transform=ax.transAxes,
        ha='right', va='top', fontsize=9,
        bbox=dict(boxstyle='round,pad=0.5', fc='white', ec='lightgray', lw=0.8))

ax.text(0.02, 0.04, 'Bubble size ∝ number of sentences that year',
        transform=ax.transAxes, fontsize=8, color='dimgray', style='italic')

ax.set_xlabel('Sentencing Year', fontsize=11)
ax.set_ylabel('Cohort Execution Rate', fontsize=11)
ax.set_title('Simple OLS: Sentencing Year vs. Cohort Execution Rate (1976–2010)',
             fontsize=13, fontweight='normal', pad=12)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f'{y:.0%}'))
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(color='lightgray', linewidth=0.6, zorder=0)
ax.legend(fontsize=9, frameon=False, loc='upper center',
          bbox_to_anchor=(0.5, -0.12), ncol=2)
plt.tight_layout()
plt.show()
````

A simple ordinary least squares (OLS) regression was fit to examine whether sentencing
year predicts the proportion of a cohort that was ultimately executed. The analysis was
restricted to sentencing cohorts from 1976 to 2010, as more recent cohorts have not had
sufficient time for cases to resolve and would artificially deflate the execution rate. Each point represents one sentencing year, with bubble size
proportional to the number of sentences issued that year.

The model reveals a strong negative relationship: for each additional year, the cohort
execution rate decreases by approximately 0.69 percentage points (β = −0.0069, p < 0.001,
95% CI [−0.0086, −0.0052]). The model explains 66.6% of the variance in cohort execution
rates (R² = 0.666), indicating that sentencing year is a strong predictor of whether a
cohort will eventually be executed. This finding is consistent with the broader decline

#### Multiple OLS — Race Coefficient Before and After Controlling for State Execution Rate

````python
from scipy import stats
from matplotlib.lines import Line2D

df['race_clean'] = df['race'].str.strip().str.title().replace({
    'Asian': 'Other', 'Native American': 'Other', 'Other Or Unknown': 'Other'
})

df['race_White']    = (df['race_clean'] == 'White').astype(float)
df['race_Black']    = (df['race_clean'] == 'Black').astype(float)
df['race_Hispanic'] = (df['race_clean'] == 'Hispanic Or Latino').astype(float)
df['race_Other']    = (df['race_clean'] == 'Other').astype(float)


if df['executed'].dtype == 'bool':
    df['executed_num'] = df['executed'].astype(int)

state_exec = df.groupby('state')['executed_num'].mean().reset_index()
state_exec.columns = ['state', 'state_exec_rate']


cols_to_drop_from_df = ['state_exec_rate_x', 'state_exec_rate_y', 'state_exec_rate']
df = df.drop(columns=[col for col in cols_to_drop_from_df if col in df.columns])

df = df.merge(state_exec, on='state', how='left')

def ols(X, y):
    X = np.column_stack([np.ones(len(X)), X.astype(float)])
    y = y.astype(float)
    beta = np.linalg.lstsq(X, y, rcond=None)[0]
    resid = y - X @ beta
    n, k = X.shape
    s2 = (resid @ resid) / (n - k)
    se = np.sqrt(s2 * np.linalg.inv(X.T @ X).diagonal())
    t_crit = stats.t.ppf(0.975, df=n-k)
    p_vals = 2 * stats.t.sf(np.abs(beta / se), df=n-k)
    return beta[1:], se[1:], p_vals[1:], beta[1:] - t_crit*se[1:], beta[1:] + t_crit*se[1:]

race_vars   = ['race_White', 'race_Black', 'race_Hispanic', 'race_Other']
race_labels = ['White', 'Black', 'Hispanic\nor Latino', 'Other']
results = {}
for rv in race_vars:
    d1 = df[['executed_num', rv]].dropna()
    d2 = df[['executed_num', rv, 'state_exec_rate']].dropna()
    b1, se1, p1, lo1, hi1 = ols(d1[[rv]].values, d1['executed_num'].values)
    b2, se2, p2, lo2, hi2 = ols(d2[[rv, 'state_exec_rate']].values, d2['executed_num'].values)
    results[rv] = dict(unadj_b=b1[0], unadj_lo=lo1[0], unadj_hi=hi1[0],
                       adj_b=b2[0],   adj_lo=lo2[0],   adj_hi=hi2[0])

fig, ax = plt.subplots(figsize=(9, 6))
y_positions = np.array([0, 1.4, 2.8, 4.2])
offset = 0.25

for i, (rv, label) in enumerate(zip(race_vars, race_labels)):
    r = results[rv]
    y_u, y_a = y_positions[i] + offset, y_positions[i] - offset

    ax.plot([r['unadj_lo'], r['unadj_hi']], [y_u, y_u], color='steelblue', linewidth=2)
    ax.plot(r['unadj_b'], y_u, 'o', color='steelblue', markersize=7, zorder=4)

    ax.plot([r['adj_lo'], r['adj_hi']], [y_a, y_a], color='darkorange', linewidth=2)
    ax.plot(r['adj_b'], y_a, 's', color='darkorange', markersize=7, zorder=4)

    ax.text(r['unadj_hi'] + 0.003, y_u, f"β={r['unadj_b']:.3f}", va='center', fontsize=8.5, color='steelblue')
    ax.text(r['adj_hi']   + 0.003, y_a, f"β={r['adj_b']:.3f}",   va='center', fontsize=8.5, color='darkorange')

ax.set_yticks(y_positions)
ax.set_yticklabels(race_labels, fontsize=11)
ax.axvline(0, color='black', linewidth=0.8, linestyle='--', alpha=0.5)

legend_elements = [
    Line2D([0],[0], color='steelblue',  marker='o', linewidth=2, markersize=7, label='Unadjusted (race only)'),
    Line2D([0],[0], color='darkorange', marker='s', linewidth=2, markersize=7, label='Adjusted (+ state execution rate)'),
]
ax.legend(handles=legend_elements, fontsize=9, frameon=False,
          loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2)

ax.set_xlabel('OLS Coefficient (β) for Race on Execution Outcome\nwith 95% Confidence Interval', fontsize=10)
ax.set_title('Multiple OLS: Race Coefficient Before and After\nControlling for State Execution Rate',
             fontsize=13, fontweight='normal', pad=12)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.grid(axis='x', color='lightgray', linewidth=0.6)
ax.set_ylim(-0.8, 4.9)
plt.tight_layout()
plt.show()
````

To examine whether geographic differences between states confound the relationship between
race and execution outcome, OLS regression models were fit for each racial group twice:
once with race as the only predictor (unadjusted), and once adding state execution rate
as a control variable (adjusted).

White defendants show a positive association with execution in both models (β = 0.059
unadjusted, β = 0.063 adjusted), meaning White individuals are more likely to have been
executed compared to the reference group, and this effect persists after controlling for
state. Black defendants show a consistent negative coefficient in both models (β = −0.046),
indicating lower execution rates relative to White defendants regardless of state.
Hispanic/Latino defendants show a notable shift where the unadjusted coefficient is near zero and non-significant, but becomes more negative and significant after adjusting for state (β = −0.042), suggesting that state explains part of their apparent similarity to White
execution rates. The "Other" group shows the largest shift between models (β = −0.093
unadjusted to β = −0.039 adjusted), indicating that state execution rate does partially
confound the relationship for this group. Overall, state execution rate acts as a
meaningful confounder for some groups but not others, highlighting the complex interplay
between geography and race in death row outcomes.

````python
slope_l, intercept_l, r_l, p_l, se_l = stats.linregress(d['log_age'], d['executed_num'])
n = len(d)

fig, ax = plt.subplots(figsize=(6, 4))
jitter = np.random.default_rng(42).uniform(-0.015, 0.015, n)
ax.scatter(d['log_age'], d['executed_num'] + jitter,
           alpha=0.15, s=8, color='dimgray', zorder=2)

x_fit  = np.linspace(d['log_age'].min(), d['log_age'].max(), 200)
y_fit  = intercept_l + slope_l * x_fit
x_mean = d['log_age'].mean()
ss_x   = ((d['log_age'] - x_mean)**2).sum()
t_crit = stats.t.ppf(0.975, df=n-2)
se_fit = se_l * np.sqrt(1/n + (x_fit - x_mean)**2 / ss_x)

ax.plot(x_fit, y_fit, color='red', linewidth=2, zorder=3, label='OLS line')
ax.fill_between(x_fit, y_fit - t_crit*se_fit, y_fit + t_crit*se_fit,
                color='red', alpha=0.15, label='95% CI')

ax.text(0.03, 0.97, f'β = {slope_l:.4f}\np = {p_l:.4f}\nR² = {r_l**2:.4f}',
        transform=ax.transAxes, ha='left', va='top', fontsize=9,
        bbox=dict(boxstyle='round,pad=0.4', fc='white', ec='lightgray', lw=0.7))
ax.legend(fontsize=9, frameon=False)
ax.set_title('OLS: log(Age) → Execution (transformed model)', fontsize=11, fontweight='normal')
ax.set_xlabel('log(Age at Sentencing)', fontsize=10)
ax.set_ylabel('Executed (0/1)', fontsize=10)
ax.set_yticks([0, 1])
ax.set_yticklabels(['No', 'Yes'], fontsize=9)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
````

## Conclusion and Discussion

This project examined what demographic, geographic, and temporal factors are associated with death row outcomes among 7,333 individuals sentenced to death in the United States between 1976 and 2019. Across a range of statistical methods including correlation analysis, independent samples t-tests, simple and multiple OLS regression, logistic regression, log transformation, a chi-square test of independence, and geographic visualization, a consistent set of findings emerged. Sentencing year was the strongest predictor of execution outcome, with cohort execution rates declining by approximately 0.69 percentage points per year (beta = -0.0069, p < .001, R² = .667) for cohorts sentenced between 1976 and 2010. State execution rate was the strongest geographic predictor, with the logistic regression identifying it as the single largest positive coefficient in the model (beta ≈ 1.20), a pattern made visually clear by the choropleth map showing dramatic variation across states. Among demographic variables, race and gender showed statistically significant but modest associations with execution: White defendants had higher execution rates relative to other racial groups, male defendants were executed at nearly twice the rate of female defendants (37.0% vs. 20.0%), and executed individuals tended to be sentenced at a slightly younger average age (29.7 years) than those not executed (31.5 years).

Taken together, the results paint a picture of capital punishment as a practice shaped more by institutional and structural forces than by individual characteristics. The steep decline in cohort execution rates over time reflects broader societal and legal shifts including declining public support for the death penalty, an increasing number of states abolishing or suspending it, and growing scrutiny of wrongful convictions. The dominance of resentencing (32.3%) over execution (19.7%) as the most common resolved outcome suggests that the appeals process, rather than execution, has become the modal endpoint for death-sentenced individuals. The large variation across states visible in both the map and the regression results reinforces the view that capital punishment in the United States is not a uniform federal policy but a patchwork of highly divergent local practices where geography can matter as much as, or more than, individual demographic characteristics.

On the question of race specifically, the project uncovered meaningful but incomplete findings. Race was statistically significant across multiple analyses, but the direction and magnitude of its association with execution shifted depending on whether geographic context was accounted for. The correlation analysis initially suggested White defendants were less likely to be executed, but the multiple OLS regression controlling for state reversed this pattern, revealing a positive association for White defendants once state execution rate was held constant. This Simpson's Paradox-type confound, where White defendants are concentrated in higher-executing states, means the raw racial pattern obscures rather than reveals the underlying relationship. Hispanic/Latino defendants showed a similar geographic confound, appearing near the baseline execution rate when unadjusted but becoming significantly less likely to be executed after controlling for state. What the data cannot tell us, however, is whether these patterns reflect racial bias in how cases are prosecuted, sentenced, or appealed, because the dataset contains only defendant race and not victim race, crime type, or legal representation quality. Race-based inferences drawn from this analysis should therefore be understood as descriptive associations rather than evidence of causal racial disparity. The project establishes that race is entangled with geography in ways that make simple conclusions about racial bias impossible to draw from this data alone.

Several important limitations constrain the confidence we can place in these findings. The most significant is missing data: approximately 83% of individuals lacked date of birth information, restricting all age-related analyses to roughly 17% of the sample (n ≈ 1,254). It is unclear whether this missingness is random or systematically related to outcome, and if records for earlier, less-documented cases are disproportionately missing, the age analyses may not generalize to the full population. Second, the dataset captures only individuals who received a death sentence and does not include people charged with capital crimes who were not sentenced to death, making it impossible to analyze earlier stages of the capital punishment pipeline such as charging decisions or plea bargains. Third, the dataset captures a snapshot in time, so individuals sentenced in more recent years have not had enough time for their cases to fully resolve, which artificially deflates observed execution rates for post-2010 cohorts. We addressed this by restricting the OLS trend analysis to cohorts through 2010, but the issue remains relevant for any inference about contemporary sentencing practices. Finally, the absence of victim characteristics, crime type, and appeal-level legal representation data is especially limiting for race-based analysis, since research consistently shows these variables interact with defendant race in shaping capital outcomes.

If this work were to continue, the most important next step for the race-based research question would be linking this dataset to victim-level data, particularly victim race, which prior single-state studies suggest is one of the strongest predictors of who gets executed. Without that variable, it is not possible to determine whether the racial patterns observed here reflect defendant-based disparities, victim-based disparities, or both. Incorporating county-level data would also enable finer-grained geographic analysis, since execution rates vary substantially by county of prosecution within states. A survival analysis framework that models time from sentencing to outcome as a function of covariates would be more statistically appropriate than the binary outcome approaches used here. Finally, as the dataset ages and more pending cases resolve, re-running these analyses on a more complete dataset would allow for more reliable estimation of true long-run execution rates by race, state, and sentencing cohort.

## Prepare Age and Execution Data

### Subtask:
Load the cleaned dataset and ensure the 'executed' column is correctly encoded as numeric (0/1) and 'age_at_sentencing' is available. Filter out any rows where 'age_at_sentencing' or 'executed' are missing, as these are crucial for the analysis. Separate the data into two groups: individuals who were 'executed' and those who were 'not executed'.


**Reasoning**:
First, I need to load the specified CSV file into a DataFrame. Then, I will convert the 'executed' column to a numeric 0/1 representation, as it's critical for the subsequent filtering and group separation. After that, I will drop rows with missing values in 'age_at_sentencing' or the newly created 'executed_numeric' column. Finally, I will create two new DataFrames based on the 'executed_numeric' column: 'executed_group' and 'not_executed_group'.



````python
file_path = "data/02-processed/condemned_cleaned.csv"
df_analysis = pd.read_csv(file_path)

# Ensure 'executed' column is correctly encoded as numeric (0/1)
# It might be loaded as object if it contains 'True'/'False' strings.
df_analysis['executed_numeric'] = df_analysis['executed'].astype(str).str.lower().map({'true': 1, 'false': 0})

# Drop rows where 'age_at_sentencing' or 'executed_numeric' are missing
initial_rows_count = len(df_analysis)
df_analysis = df_analysis.dropna(subset=['age_at_sentencing', 'executed_numeric']).copy()
print(f"Removed {initial_rows_count - len(df_analysis)} rows due to missing 'age_at_sentencing' or 'executed_numeric'.")

# Ensure 'executed_numeric' is integer type after dropping NaNs, if it wasn't already.
df_analysis['executed_numeric'] = df_analysis['executed_numeric'].astype(int)

# Separate data into executed and not executed groups
executed_group = df_analysis[df_analysis['executed_numeric'] == 1].copy()
not_executed_group = df_analysis[df_analysis['executed_numeric'] == 0].copy()

print(f"Shape of df_analysis after filtering: {df_analysis.shape}")
print(f"Shape of executed_group: {executed_group.shape}")
print(f"Shape of not_executed_group: {not_executed_group.shape}")

print("\nFirst 5 rows of executed_group:")
print(executed_group[['full_name', 'age_at_sentencing', 'executed', 'executed_numeric']].head())

print("\nFirst 5 rows of not_executed_group:")
print(not_executed_group[['full_name', 'age_at_sentencing', 'executed', 'executed_numeric']].head())
````

**Reasoning**:
The subtask requires performing an independent samples Welch's t-test and then visualizing the age distributions using Kernel Density Estimates (KDE) for both executed and non-executed groups. I will also add the t-test statistics to the plot.



````python
import matplotlib.pyplot as plt
from scipy import stats
from scipy.stats import gaussian_kde
import numpy as np

# Extract age data for executed and not executed groups
executed_ages = executed_group['age_at_sentencing']
not_executed_ages = not_executed_group['age_at_sentencing']

# Perform independent samples Welch's t-test
t_statistic, p_value = stats.ttest_ind(executed_ages, not_executed_ages, equal_var=False)

# Create the KDE plot
fig, ax = plt.subplots(figsize=(10, 6))

# Plot KDE for executed group
kde_executed = gaussian_kde(executed_ages, bw_method=0.35)
x_range = np.linspace(10, 80, 300) # Assuming age range 10-80
ax.plot(x_range, kde_executed(x_range), color='red', linestyle='-', linewidth=2, label=f'Executed (n={len(executed_ages):,})')
ax.axvline(executed_ages.mean(), color='red', linestyle=':', linewidth=1.4, alpha=0.8)

# Plot KDE for not executed group
kde_not_executed = gaussian_kde(not_executed_ages, bw_method=0.35)
ax.plot(x_range, kde_not_executed(x_range), color='steelblue', linestyle='--', linewidth=2, label=f'Not Executed (n={len(not_executed_ages):,})')
ax.axvline(not_executed_ages.mean(), color='steelblue', linestyle=':', linewidth=1.4, alpha=0.8)

# Annotate means
ymax = ax.get_ylim()[1]
ax.text(executed_ages.mean() - 1.2, ymax * 0.22, f'mean={executed_ages.mean():.1f}',
        color='red', fontsize=9.5, ha='right')
ax.text(not_executed_ages.mean() + 0.8, ymax * 0.08, f'mean={not_executed_ages.mean():.1f}',
        color='steelblue', fontsize=9.5, ha='left')

# Add t-test results to the plot
stats_text = f't({len(executed_ages) + len(not_executed_ages) - 2:.0f}) = {t_statistic:.3f}\np = {p_value:.4f}'
ax.text(0.02, 0.97, stats_text, transform=ax.transAxes,
        ha='left', va='top', fontsize=10,
        bbox=dict(boxstyle='round,pad=0.5', fc='white', ec='lightgray', lw=0.8))

# Set plot labels and title
ax.set_xlabel('Age at Sentencing (years)', fontsize=11)
ax.set_ylabel('Density', fontsize=11)
ax.set_title('Distribution of Age at Sentencing by Execution Outcome', fontsize=13, fontweight='normal', pad=12)
ax.legend(fontsize=10, frameon=False)

# Customize plot appearance
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(axis='y', color='lightgray', linewidth=0.6, zorder=0)

plt.tight_layout()
plt.show()

print(f"\nWelch's t-test results: T-statistic = {t_statistic:.3f}, P-value = {p_value:.4f}")
print(f"Mean age at sentencing for executed individuals: {executed_ages.mean():.1f} years")
print(f"Mean age at sentencing for non-executed individuals: {not_executed_ages.mean():.1f} years")
````
