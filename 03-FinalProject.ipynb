## Background and Prior Work

Our interest in research is on what factors are associated with death row outcomes among individuals sentenced to death in the United States since 1976, including demographics, legal factors, and geographic distribution. The death penalty in the United States remains one of the most researched topics of the criminal justice system, with outcomes on death row shaped by more than just legal facts. Since its reestablishment in 1976, thousands of individuals have been sentenced to death, executed, exonerated, had sentences commuted, or died from other causes while on death row<sup><a name="ref3"></a>3</sup>. National data from *Death Row U.S.A.* indicate that death row populations are disproportionately male (98.9 %) and primarily White or Black, with significant variation across states, suggesting the importance of geographic context in analyzing death row outcomes<sup><a name="ref3"></a>3</sup>.

Research on specific jurisdictions has studied how demographic factors, particularly race and ethnicity, influence outcomes after a death sentence. Petrie and Coverdill researched Texas death row cases from 1974 to 2009 and found that the race of both offenders and victims affected the likelihood of execution, while having less impact on sentence relief such as commutation or overturning<sup><a name="ref5"></a>5</sup>. These findings suggest that while executions may reflect demographic disparities, other outcomes are shaped by different factors, including appeals processes and legal context.

The results in studies of high-activity death penalty counties show how race interacts with legal processes. Scott Phillips’ analysis of Harris County, Texas, shows that both defendant and victim race influenced whether a defendant was charged with capital murder and ultimately given the death penalty<sup><a name="ref4"></a>4</sup>. Combined with national data, these studies indicate that death row outcomes arise from the intersection of individual characteristics, legal decision-making, and geographic variation. Understanding these factors is important when addressing questions about individual justice and disparities in the administration of capital punishment.

### Data Overview — The Condemned: Death Sentences in the Modern Era (1976–2019)

- Dataset Name: The Condemned  
- Link to the dataset: https://raw.githubusercontent.com/theintercept/the-condemned-data/refs/heads/master/the-condemed-data.csv  
- Number of observations: 7,335

This dataset, compiled by *The Intercept*, includes all known individuals sentenced to death in active U.S. death penalty jurisdictions from July 2, 1976 (the date of *Gregg v. Georgia*, which reinstated the modern era of capital punishment) through November 30, 2019. Each row represents one individual sentenced to death during the modern death penalty era.

As of 2019:
- 1,448 individuals were executed  
- 2,752 remain on death row  
- 3,135 were removed from death row for reasons other than execution (such as commutations, exonerations, resentencing, death in custody, or release)

This dataset allows us to examine how demographic, geographic, and temporal factors are associated with different death row outcomes across jurisdictions in the modern U.S. death penalty system.

### Variables most relevant to this project

Outcome variables
- executed: Indicates whether the individual was executed.
- volunteer_execution: Indicates whether the individual waived appeals and voluntarily accepted execution.
- commuted: Indicates whether the death sentence was commuted through executive action.
- exonerated: Indicates whether the individual was legally cleared of the charges.
- resentenced: Indicates whether the individual received a non-death sentence after review.
- released: Indicates whether the individual was released from prison.
- died / suicide: Indicates whether the individual died (including suicide) while on death row.
- currently_on_death_row: Indicates whether the individual remained on death row at the time of the dataset’s most recent update.

Demographic variables
- race: Recorded race of the individual, used to examine potential racial disparities in death row outcomes.
- gender: Recorded gender of the individual.
- DOB (date of birth): Used to calculate age at the time of sentencing.

Geographic variables
- state: The U.S. state where the individual was sentenced to death, allowing comparison across jurisdictions.
- county: The county of sentencing, reflecting the local legal and judicial context.

Additional variables
- sentencing_year: The year the death sentence was issued, allowing analysis of trends in capital punishment practices over time.
- current_sentence: Describes the individual’s current legal status after removal from death row or continued incarceration. Possible values include DR (Death Row), LWOP (Life Without Parole), LPP (Life With Possibility of Parole), Y (Term of Years), M (Term of Months), and TIME (Time Served).

### Dataset limitations relevant to this project

1. The dataset only includes individuals who received a death sentence. It does not include people charged with capital crimes who did not receive the death penalty, so disparities in who receives a death sentence cannot be analyzed.

2. The data were compiled from corrections departments, public defenders, advocacy groups, and independent trackers. Reporting standards may vary across states and may introduce inconsistencies.

3. Individuals sentenced earlier have had more time for appeals, commutations, or execution, so recent sentences may appear more often as “currently on death row.”

4. The dataset does not include detailed information about appeals, crime type, victim characteristics, or offense severity, which are important legal factors that may influence sentencing outcomes.

5. The dataset also lacks specific dates for executions, commutations, or exonerations, making it difficult to measure time spent on death row or analyze how quickly different outcomes occur.

## Results

### Exploratory Data Analysis

#### Death Sentences and Executions Over Time

This figure examines temporal trends in death sentencing and execution rates from 1976 to 2019. 
Understanding how the number of new death sentences and executions have changed over time is important context 
for interpreting demographic patterns, since individuals sentenced in earlier years have had more time for their cases to resolve.

```python
sentences_by_year = df.groupby('sentencing_year').size().reset_index(name='sentences')

# Executions per year sentenced (as a proportion of sentences that year)
exec_by_year = df[df['executed'] == True].groupby('sentencing_year').size().reset_index(name='executions')
trend = sentences_by_year.merge(exec_by_year, on='sentencing_year', how='left').fillna(0)
trend['execution_rate'] = trend['executions'] / trend['sentences']

fig, ax1 = plt.subplots(figsize=(12, 5))

ax1.bar(trend['sentencing_year'], trend['sentences'], color='steelblue', alpha=0.5, label='New Death Sentences')
ax1.set_xlabel('Year of Sentencing')
ax1.set_ylabel('Number of New Death Sentences', color='steelblue')
ax1.tick_params(axis='y', labelcolor='steelblue')

ax2 = ax1.twinx()
ax2.plot(trend['sentencing_year'], trend['execution_rate'], color='crimson', linewidth=2, label='Execution Rate')
ax2.set_ylabel('Execution Rate (proportion of cohort)', color='crimson')
ax2.tick_params(axis='y', labelcolor='crimson')
ax2.set_ylim(0, 1)

plt.title('Death Sentences per Year and Eventual Execution Rate by Sentencing Cohort')
fig.legend(loc='upper right', bbox_to_anchor=(0.88, 0.88))
plt.tight_layout()
plt.show()
```

Death sentencing peaked in the mid-1990s, reaching nearly 300 new sentences per year, before declining sharply through the 2000s and into the 2010s. By 2019, fewer than 50 new sentences were issued annually. The execution rate (red line) remained relatively stable between roughly 20–30% for cohorts sentenced from the late 1970s through the mid-1990s, then drops steeply for cohorts sentenced after approximately 2003, approaching zero for the most recent years.

To explore the relationships between demographic and temporal variables and death row outcomes, we compute point-biserial correlations between each predictor and the binary execution outcome. This provides a quantitative summary of which variables are most strongly associated with whether an individual was ultimately executed.

```python
# Encode execution as numeric
df['executed_num'] = df['executed'].map({True: 1, False: 0})

# Encode race as dummy variables
df['race_clean'] = df['race'].str.strip().str.title().replace({
    'Asian': 'Other', 'Native American': 'Other', 'Other Or Unknown': 'Other'
})
race_dummies = pd.get_dummies(df['race_clean'], prefix='race')
df_encoded = pd.concat([df, race_dummies], axis=1)

# Encode gender
df_encoded['gender_male'] = (df_encoded['gender'].str.strip().str.lower() == 'male').astype(int)

# State-level execution rate (proportion of sentences in that state that were executed)
state_exec_rate = df.groupby('state')['executed_num'].mean().reset_index()
state_exec_rate.columns = ['state', 'state_exec_rate']
df_encoded = df_encoded.merge(state_exec_rate, on='state', how='left')

# Variables to correlate with execution
corr_vars = ['sentencing_year', 'gender_male', 'state_exec_rate',
             'race_Black', 'race_Hispanic Or Latino', 'race_White']

corr_results = {}
for var in corr_vars:
    valid = df_encoded[['executed_num', var]].dropna()
    r, p = stats.pointbiserialr(valid['executed_num'], valid[var])
    corr_results[var] = {'r': round(r, 3), 'p': round(p, 4)}

corr_df = pd.DataFrame(corr_results).T.sort_values('r', ascending=False)
corr_df.index = ['Sentencing Year', 'Gender (Male)', 'State Execution Rate',
                 'Race: Black', 'Race: Hispanic/Latino', 'Race: White']

fig, ax = plt.subplots(figsize=(9, 4))
colors = ['crimson' if r > 0 else 'steelblue' for r in corr_df['r']]
bars = ax.barh(corr_df.index, corr_df['r'], color=colors, alpha=0.8)
ax.axvline(0, color='black', linewidth=0.8)
ax.set_xlabel('Point-Biserial Correlation (r) with Execution Outcome')
ax.set_title('Correlation of Predictor Variables with Execution Outcome')

for bar, (_, row) in zip(bars, corr_df.iterrows()):
    sig = '***' if row['p'] < 0.001 else ('**' if row['p'] < 0.01 else ('*' if row['p'] < 0.05 else ''))
    ax.text(bar.get_width() + 0.003, bar.get_y() + bar.get_height()/2,
            f"r={row['r']:.3f} {sig}", va='center', fontsize=9)

ax.text(0.01, -0.12, '* p<.05  ** p<.01  *** p<.001', transform=ax.transAxes, fontsize=6, color='gray')
plt.tight_layout()
plt.show()

print('\nCorrelation Summary Table:')
print(corr_df.to_string())
```

The correlation analysis shows that **sentencing year** has the strongest association with execution outcome (r = 0.482, p < .001). This indicates that individuals sentenced in more recent years are more likely to appear as executed in the dataset. **Gender (Male)** shows a small but statistically significant positive correlation with execution (r = 0.074, p < .001), suggesting males are slightly more likely to have been executed. **State execution rate** also shows a small positive correlation (r = 0.031, p < .01).

All three racial categories show negative correlations with execution. **Race: White** has the largest negative correlation (r = −0.189, p < .001), followed by **Race: Hispanic/Latino** (r = −0.056, p < .001). **Race: Black** shows a correlation close to zero and is not statistically significant (r = −0.010, p = .37). Overall, these results suggest that demographic variables have relatively small associations with execution outcomes, and the strongest relationship observed in this exploratory analysis is related to **sentencing year**.

#### Logistic Regression: Predicting Execution Outcome

To quantify how multiple demographic and geographic variables jointly relate to execution outcome, we fit a logistic regression model. This allows us to examine the contribution of each variable while controlling for the others, providing a more interpretable picture of the factors associated with execution.

```python
# Build model dataset
model_vars = ['sentencing_year', 'gender_male', 'state_exec_rate',
              'race_Black', 'race_Hispanic Or Latino', 'race_White']

model_df = df_encoded[model_vars + ['executed_num']].dropna()

X = model_df[model_vars]
y = model_df['executed_num']

# Standardize for interpretable coefficients
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Fit logistic regression
lr = LogisticRegression(max_iter=1000, random_state=42)
lr.fit(X_scaled, y)

# Coefficient plot
coef_labels = ['Sentencing Year', 'Gender (Male)', 'State Exec. Rate',
               'Race: Black', 'Race: Hispanic/Latino', 'Race: White']
coefs = lr.coef_[0]

# Sort by magnitude
sorted_idx = np.argsort(coefs)
coefs_sorted = coefs[sorted_idx]
labels_sorted = [coef_labels[i] for i in sorted_idx]
colors = ['crimson' if c > 0 else 'steelblue' for c in coefs_sorted]

fig, ax = plt.subplots(figsize=(9, 4))
ax.barh(labels_sorted, coefs_sorted, color=colors, alpha=0.8)
ax.axvline(0, color='black', linewidth=0.8)
ax.set_xlabel('Standardized Logistic Regression Coefficient\n(positive = more likely to be executed)')
ax.set_title('Logistic Regression: Variables Associated with Execution Outcome')

# Accuracy
acc = lr.score(X_scaled, y)
ax.text(0.98, 0.02, f'Model Accuracy: {acc:.2%}', transform=ax.transAxes,
        ha='right', fontsize=9, color='gray')

plt.tight_layout()
plt.show()

print(classification_report(y, lr.predict(X_scaled)))
```

The logistic regression model (accuracy: 82.76%) confirms that **state execution rate** is the strongest predictor (coefficient ≈ 1.20), indicating that geographic differences across states are strongly associated with execution outcomes. **Sentencing year** has the largest negative coefficient (≈ −0.60), suggesting that individuals sentenced in more recent years are less likely to have been executed within the timeframe captured by the dataset.

Among demographic variables, **Race: White** has the second largest positive coefficient (≈ 0.35), followed by **Race: Black** and **Gender (Male)** (≈ 0.12 each), while **Race: Hispanic/Latino** is close to zero. The model has difficulty predicting executions because the dataset is imbalanced, with 1,446 executed cases compared to 5,876 non-executed cases. As a result, the model shows lower precision (0.60) and recall (0.38) for the executed class. These results should be interpreted as exploratory associations rather than predictive conclusions.
